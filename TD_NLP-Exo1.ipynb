Select 5 positive and 5 negative reviews from the dataset.

● Build a unique vocabulary list from these 10 reviews.

● For each review, create a BoW vector showing the frequency of each vocabulary word in the review.

● Question: Which words are most frequent in positive vs. negative
reviews?
 What does this tell you about the sentiment features captured
by BoW?

```python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer

#Load dataset
df = pd.read_csv('/content/IMDB Dataset.csv')
df.head()
```

```python
 #shape
 #Select positive and negative reviews
df_positive =df[df["sentiment"]=="positive"]
df_negative =df[df["sentiment"]=="negative"]
```

```python
df_postive_5=df_positive.head()
df_negative_5=df_negative.head()

df_10=pd.concat([df_postive_5,df_negative_5])
#build a sentence by concateneting all words in the df
corpus=''.join(df_10["review"].astype(str))
```

```python
import nltk
import re
from nltk.tokenize import word_tokenize
```

```python
nltk.download('punkt')
nltk.download('punkt_tab')
```

```python
#Tokens creation
tokens = word_tokenize(corpus)
```

```python
#regex to choose only alphabetical words
#re.sub(r"[^a-zA-Z0-9\s]", "", text)
clean_corpus=re.sub(r'[^a-zA-Z\s]','',corpus.lower())
```

```python
from sklearn.feature_extraction.text import CountVectorizer
#Bag of words
vectorizer = CountVectorizer(stop_words="english")
#Mehode 1
#Fit transform on all lines
X = vectorizer.fit_transform([clean_corpus])
#Methode 2 fit transform on each line
```

```python
df_bofw=pd.DataFrame(X.toarray(),columns=vectorizer.get_feature_names_out())
```

```python
df_bofw['sentiment']=df_10['sentiment']
```

```python
#build a sentence by concateneting all words in the df
df_10['clean_text']=df_10['review'].apply(lambda x : re.sub(r'[^a-zA-Z\s]','',str(x).lower()))
```

```python
vectorizer = CountVectorizer(stop_words="english")
#Mehode 1
#Fit transform on all lines
X = vectorizer.fit_transform(df_10['clean_text'])#texte total des 10 lignesb
```

```python
vectorizer.fit(df_10['clean_text']) #train on all lines
```

```python
vectorizer.transform(df_10['clean_text']) #predict on each line
```

```python
df_bofw=pd.DataFrame(X.toarray(),columns=vectorizer.get_feature_names_out())
df_bofw['sentiment']=df_10['sentiment']
```

```python
#Pos_df
df_pos=df_bofw[df_bofw['sentiment']=='positive'].drop(columns=['sentiment'])
#neg_df
df_neg=df_bofw[df_bofw['sentiment']=='negative'].drop(columns=['sentiment'])
```
