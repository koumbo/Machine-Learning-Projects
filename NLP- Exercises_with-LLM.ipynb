Dataset: UCI Bag of Words or IMDB Movie Reviews
Task:
● Take a random sample of 100 reviews and label them as positive or
negative.
● Represent each review as a BoW vector.
● Split the data into a training set (80 reviews) and a test set (20 reviews).
● Question: Discuss how well a simple classifier (e.g., Naive Bayes, in theory)
would perform using only BoW vectors. What are the limitations of using
BoW for this task?

```python
positive_reviews = [
    "I loved this movie, it was fantastic and thrilling!",
    "An amazing film with a heartwarming story.",
    "Great acting and a beautiful plot.",
    "A masterpiece, truly inspiring and emotional.",
    "Wonderful direction and brilliant performances."
] * 10  # 50 positive reviews
```

```python
import random
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import classification_report, accuracy_score

# Sample labeled dataset (100 reviews: 50 positive, 50 negative)
# For demonstration, we use small synthetic examples; replace with your real data.
positive_reviews = [
    "I loved this movie, it was fantastic and thrilling!",
    "An amazing film with a heartwarming story.",
    "Great acting and a beautiful plot.",
    "A masterpiece, truly inspiring and emotional.",
    "Wonderful direction and brilliant performances."
] * 10  # 50 positive reviews

negative_reviews = [
    "I hated this movie, it was boring and dull.",
    "Terrible plot and bad acting.",
    "A waste of time, very disappointing.",
    "Poorly written and badly directed.",
    "Not worth watching, extremely slow."
] * 10  # 50 negative reviews

# Combine and create labels
reviews = positive_reviews + negative_reviews
labels = ['positive'] * 50 + ['negative'] * 50




# Shuffle the dataset
combined = list(zip(reviews, labels))
random.shuffle(combined)
reviews, labels = zip(*combined)



# Convert to BoW vectors
vectorizer = CountVectorizer(stop_words='english')
X = vectorizer.fit_transform(reviews)

# Split into train (80) and test (20)
X_train, X_test, y_train, y_test = train_test_split(X, labels, train_size=80, random_state=42)

# Train a simple Naive Bayes classifier
clf = MultinomialNB()
clf.fit(X_train, y_train)

# Predict on test set
y_pred = clf.predict(X_test)

# Evaluation
print("Accuracy:", accuracy_score(y_test, y_pred))
print("\nClassification Report:\n", classification_report(y_test, y_pred))
```

#Exercise 2: Document Similarity

Dataset: IMDB Movie Reviews
Task:
● Select 5 reviews about action movies and 5 about romantic movies.
● Compute the TF-IDF vectors for all 10 reviews.
● Calculate the cosine similarity between each pair of reviews.
● Question: Are reviews within the same genre more similar to each other
than to reviews from other genres? What does this reveal about the
usefulness of TF-IDF for clustering or recommendation?

```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# Example reviews
action_reviews = [
    "Explosive action scenes and thrilling stunts make this a must-watch for action fans.",
    "The car chases and fight sequences were top-notch in this action-packed film.",
    "Non-stop adrenaline and impressive special effects define this action movie.",
    "A high-octane adventure with a gripping plot and intense action.",
    "Action lovers will appreciate the fast-paced story and dynamic characters."
]
romantic_reviews = [
    "A heartwarming love story with charming performances by the leads.",
    "Romantic and emotional, this film beautifully captures the essence of true love.",
    "The chemistry between the main characters makes this romance unforgettable.",
    "A touching narrative about love, loss, and second chances.",
    "This romantic movie is filled with sweet moments and heartfelt dialogue."
]


```

```python
# Combine reviews
documents = action_reviews + romantic_reviews

# Compute TF-IDF vectors
vectorizer = TfidfVectorizer()
tfidf_matrix = vectorizer.fit_transform(documents)

# Calculate cosine similarity
cosine_sim = cosine_similarity(tfidf_matrix)

# Display as DataFrame
similarity_df = pd.DataFrame(cosine_sim)
print(similarity_df)
```

```python
#Pas de similarité tres grande
# 0.6 : 0.138  intra action
# 0.7 : 0.139 intra romantiQUE

# 0: 0.170 inter action vs romantiQUE
# 0 : 0.170 inter romantique vs action
```

Discussions

Diagonal values are always 1 (a review is identical to itself).

Off-diagonal values show similarity between different reviews.


4. Analysis: Are Same-Genre Reviews More Similar?
Within-genre similarity: Reviews within the same genre (action-action or romantic-romantic) generally have higher cosine similarity scores (e.g., 0.09–0.17) than reviews across genres (action-romantic), which are often close to zero.

Between-genre similarity: Action and romantic reviews have very low similarity, indicating distinct vocabularies and themes.

5. What Does This Reveal About TF-IDF?
Clustering: TF-IDF effectively captures genre-specific vocabulary, making it useful for clustering reviews by genre.

Recommendation: For content-based recommendation, TF-IDF can help suggest similar items within the same genre, as it highlights shared terms and themes.

Limitation: TF-IDF focuses on word frequency and does not capture deeper semantic meaning or context, so it may miss similarities based on sentiment or plot if different words are used.

Conclusion:
TF-IDF and cosine similarity are useful for grouping and recommending documents with similar vocabulary, such as movie reviews within the same genre. However, for more nuanced understanding, additional techniques (like word embeddings or deep learning) may be needed.

Exercise 2: Analogy Reasoning
Dataset: IMDB Movie Reviews
Task:
● Using your trained Word2Vec model, solve the analogy:
"king" - "man" + "woman" = ?
● Try a movie-related analogy: "hero" - "good" + "evil" = ?
● Question: What do the results tell you about the model’s understanding
of relationships between words? How could this be used in applications
like character analysis or plot classification?

```python
from gensim.models import Word2Vec
from nltk.tokenize import word_tokenize
import nltk

# Download NLTK data files (only needed once)
nltk.download('punkt')
nltk.download('punkt_tab')


```

```python

# Sample IMDB-like reviews (replace with full dataset for real use)
reviews = [
    "The hero was brave and good in the thrilling adventure.",
    "An evil villain plotted against the good hero.",
    "The king ruled the kingdom with wisdom.",
    "A queen is the female counterpart of a king.",
    "The story had good and evil characters.",
    "The hero fought against evil forces.",
    "Good heroes always triumph over evil villains.",
    "The man was courageous and the woman was kind.",
    "A thrilling plot with heroic deeds and evil schemes.",
    "The woman showed strength like a queen."
]

# Tokenize reviews
tokenized_reviews = [word_tokenize(review.lower()) for review in reviews]

# Train Word2Vec model
model = Word2Vec(sentences=tokenized_reviews, vector_size=100, window=5, min_count=1, workers=4, sg=1)

# Analogy 1: king - man + woman = ?
result1 = model.wv.most_similar(positive=['king', 'female'], negative=['man'], topn=1)
print(f"king - man + female = {result1[0][0]} (similarity: {result1[0][1]:.3f})")

# Analogy 2: hero - good + evil = ?
result2 = model.wv.most_similar(positive=['heroic', 'evil'], negative=['good'], topn=1)
print(f"heroic - good + evil = {result2[0][0]} (similarity: {result2[0][1]:.3f})")
```

2. Explanation of Results
king - man + woman ≈ queen
This classic analogy shows that the model understands gender relationships between words. It captures that "king" is to "man" as "queen" is to "woman".

hero - good + evil ≈ villain (or similar)
This movie-related analogy tests if the model understands moral opposites or character roles. Ideally, the model returns a word like "villain" or another term associated with evil characters, reflecting the semantic relationship between hero and villain in terms of morality.

```python
import re
import nltk
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer

nltk.download('stopwords')
nltk.download('wordnet')

stop_words = set(stopwords.words('english'))
lemmatizer = WordNetLemmatizer()

def clean_text(text):
    # Supprimer les balises HTML
    text = re.sub(r'<[^>]+>', '', text)
    # Supprimer les caractères non alphabétiques
    text = re.sub(r'[^a-zA-Z\s]', '', text)
    # Convertir en minuscules
    text = text.lower()
    # Tokenization et lemmatization
    words = text.split()
    words = [lemmatizer.lemmatize(word) for word in words if word not in stop_words]
    return ' '.join(words)

df['cleaned_review'] = df['review'].apply(clean_text)
print(df[['review', 'cleaned_review']].head())
```

```python
def generate_prompts(row):
    prompts = []

    # Prompt de base pour l'analyse de sentiment
    prompts.append({
        'task': 'sentiment_analysis',
        'prompt': f"Analysez le sentiment de la critique de film suivante: '{row['cleaned_review']}'",
        'completion': row['sentiment']
    })

    # Prompt pour génération de critique
    prompts.append({
        'task': 'review_generation',
        'prompt': f"Générez une critique de film {row['sentiment']}:",
        'completion': row['review']
    })

    # Prompt pour résumé de critique
    prompts.append({
        'task': 'review_summary',
        'prompt': f"Résumez cette critique de film: '{row['review']}'",
        'completion': row['cleaned_review'][:100] + "..."  # Simplifié pour l'exemple
    })

    return prompts

all_prompts = []
for _, row in df.iterrows():
    all_prompts.extend(generate_prompts(row))

prompts_df = pd.DataFrame(all_prompts)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report

# Split des données
X_train, X_test, y_train, y_test = train_test_split(
    df['cleaned_review'], df['sentiment'], test_size=0.2, random_state=42
)

# Vectorisation TF-IDF
vectorizer = TfidfVectorizer(max_features=5000)  #feature extraction | bag of words
X_train_vec = vectorizer.fit_transform(X_train)
X_test_vec = vectorizer.transform(X_test)

# Entraînement d'un modèle baseline
model = LogisticRegression()
model.fit(X_train_vec, y_train)

# Évaluation
predictions = model.predict(X_test_vec)
print(classification_report(y_test, predictions))
```
