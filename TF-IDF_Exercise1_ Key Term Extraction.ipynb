```python
import pandas as pd

# Load the dataset
df = pd.read_csv("IMDB Dataset.csv")

# Randomly sample 10 reviews
sampled_reviews = df.sample(10, random_state=42).reset_index(drop=True)
```

```python
import string

def preprocess(text):
    text = text.lower()
    text = text.translate(str.maketrans('', '', string.punctuation))
    return text

sampled_reviews['cleaned'] = sampled_reviews['review'].apply(preprocess)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

# Initialize vectorizer
vectorizer = TfidfVectorizer(stop_words='english')

# Fit and transform the cleaned reviews
X = vectorizer.fit_transform(sampled_reviews['cleaned'])

# Get the vocabulary (column names)
feature_names = vectorizer.get_feature_names_out()
```

```python
import numpy as np

top_words_per_review = []

for i in range(X.shape[0]):
    row = X[i].toarray()[0]
    top_indices = row.argsort()[-3:][::-1]  # Top 3 TF-IDF indices
    top_words = [(feature_names[idx], row[idx]) for idx in top_indices]
    top_words_per_review.append(top_words)

# Display the results
for idx, words in enumerate(top_words_per_review):
    print(f"\nReview {idx+1}:")
    for word, score in words:
        print(f"  {word}: {score:.4f}")
```
