```python
from transformers import AutoTokenizer, AutoModelForCausalLM

# Load your fine-tuned model (uploaded to Colab or accessible from Drive)
model_path = 'bit_model'  # Replace with the actual path to your model folder
tokenizer = AutoTokenizer.from_pretrained(model_path)
model = AutoModelForCausalLM.from_pretrained(model_path)
```

```python
def chat_with_model(question, context_examples=None, max_length=128):
    prompt = ''
    if context_examples:
        for ex in context_examples:
            prompt += f"Question: {ex['prompt']}\nAnswer: {ex['response']}\n\n"
    prompt += f"Question: {question}\nAnswer:"

    input_ids = tokenizer(prompt, return_tensors='pt').input_ids
    output = model.generate(input_ids, max_length=max_length, pad_token_id=tokenizer.eos_token_id)
    result = tokenizer.decode(output[0], skip_special_tokens=True)
    return result[len(prompt):].strip()
```

```python
# Example usage
question = input("Posez votre question : ")
response = chat_with_model(question)
print("Réponse du modèle :", response)
```
