```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeClassifier, plot_tree
```

```python
mode_transport = pd.read_csv('/content/table_mode_transport.csv')
```

```python
mode_transport.head()
```

Preparation des donnees


```python
X = pd.get_dummies(mode_transport[["Sexe", "Nombre_Vehicule_Possede", "Frais_de_Voyage", "Niveau_de_Revenu"]])
y = mode_transport["Mode_de_Transport"]
```

entraine l'arbre de decision

```python
clf = DecisionTreeClassifier(criterion="entropy", random_state=0)
clf.fit(X, y)
```

```python
plt.figure(figsize=(16, 8))
plot_tree(clf,
          feature_names=X.columns,
          class_names=clf.classes_,
          filled=True,
          rounded=True)
plt.title("Decision_Tree-Mode_de_transport")
plt.show()
```
